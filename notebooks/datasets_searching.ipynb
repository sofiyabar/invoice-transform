```python
from datasets import load_dataset

# Загружаем SROIE (версия с запеченным OCR и разметкой)
dataset = load_dataset("priyank-m/SROIE_2019_text_recognition", split="train")

# Посмотрим на структуру первого примера
sample = dataset[0]

# 1. Формируем сырой OCR-текст (склеиваем слова, распознанные на чеке)
raw_ocr_text = " ".join(sample["words"])

# 2. Получаем Ground Truth (ключевые сущности)
ground_truth = {
    "company": sample.get("company", ""),
    "date": sample.get("date", ""),
    "address": sample.get("address", ""),
    "total": sample.get("total", "")
}

print("--- Сырой OCR-текст (Вход) ---")
print(raw_ocr_text[:300] + "...\n")  # печатаем первые 300 символов

print("--- Ground Truth JSON (Выход) ---")
print(ground_truth)
```
